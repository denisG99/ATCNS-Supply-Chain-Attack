The goal of this analysis is defining if the detected attack must be considered benign or malicious. To do so, here we try to draw a carattererization of the attack and find some relation between user if there are

```python
import pandas as pd
```

```python
DATA_PATH = "../../data/history/dataset/test.csv"

df = pd.read_csv(DATA_PATH)

df
```

# Attack characterization

```python
def is_function(x: str) -> bool:
    for string in ["func", "afunc", "patch_decorator_import", "patch_decorator_usage", "contextmanager_import", "contextmanager_usage", "overwrite_method_class", "eval_call", "exec_call", "with_statement"]:
        if string in x:
            return True
    return False
```

```python
df["target"] = df["id_res"].apply(
    lambda x: "class" if "class_" in x
    else "variable" if "var_" in x
    else "function" if is_function(x)
    else "other"
)

df.groupby("target").size().reset_index(name="count")
```

# Users relations

```python
df_relationships = df.loc[::,["id_res", "package", "file", "who_introduce", "who_remove"]]

df_relationships.dropna(inplace=True)

df_relationships.drop_duplicates(inplace=True)

df_relationships.groupby(["who_introduce", "who_remove"]).size().reset_index(name="count")
```
